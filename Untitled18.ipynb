```python
#import pandas

import numpy as np
import pandas as pd
```

```python
#read salaries.csv file and assign it to sal

sal = pd.read_csv('Salaries.csv')
```

```python
#check sal

sal.head(2)
```

```python
#use info method and know how many entries there are

sal.info()
```

```python
#avg base pay

float(sal['BasePay'].mean())
```

```python
#highest overtime pay

float(sal['OvertimePay'].max())
```

```python
#job title of joseph driscoll

sal[sal['EmployeeName'] == 'JOSEPH DRISCOLL']['JobTitle']
```

```python
#earnings of joseph driscoll

sal[sal['EmployeeName'] == 'JOSEPH DRISCOLL']['TotalPayBenefits']
```

```python
#highest paid person
#method 1

sal[sal['TotalPayBenefits'] == sal['TotalPayBenefits'].max()]['EmployeeName']
```

```python
#method 2

sal.loc[sal['TotalPayBenefits'].idxmax()]['EmployeeName']  #we can also use argmax from numpy
```

```python
#lowest paid person and notice anything strange
#method 1

sal[sal['TotalPayBenefits'] == sal['TotalPayBenefits'].min()]['EmployeeName']
```

```python
#method 2

sal.loc[sal['TotalPayBenefits'].argmin()]
```

```python
#sal['BasePay'] = pd.to_numeric(sal['BasePay'], errors='coerce')
```

```python
#avg base of all employees per year (2011-2014)

sal.groupby('Year')['BasePay'].mean()
```

```python
#number of job titles

sal['JobTitle'].nunique()
```

```python
#Top 5 most common jobs

sal['JobTitle'].value_counts().head(5)
```

```python
#Job titles with only one occurance in 2013

sum(sal[sal['Year'] == 2013]['JobTitle'].value_counts() == 1)
```

```python
#How many people had their word chief in their title

def chief_string(title):
    if 'chief' in title.lower().split():
        return True
    else:
        return False

sum(sal['JobTitle'].apply(lambda x: chief_string(x)))
```

```python
#correlation between job title string and salary

sal['title_len'] = sal['JobTitle'].apply(len)
```

```python
sal[['JobTitle', 'title_len']].corr()
```
